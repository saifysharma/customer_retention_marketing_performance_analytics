## 3. Handling Potential Outliers

### Outlier Detection Using IQR

```python
Q1 = df['total_spent'].quantile(0.25)
Q3 = df['total_spent'].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower:", lower)
print("Upper:", upper)
```

```python
df[df['total_spent'] > upper][['customer_id', 'total_spent']]
```
